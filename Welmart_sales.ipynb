## Librerias

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

## Cargar datos

```python
df = spark.read.csv('/Volumes/workspace/trabajosia/info_trabajo/Superstore.csv', header=True, inferSchema=True)
display(df.limit(10))
```

```python
    
# Filter rows where "Sales" is a valid number
df_clean = df.filter(
    regexp_extract(col("Sales"), r'^-?\d+(\.\d+)?$', 0) != ""
)

# Cast columns after filtering
df_clean = (
    df_clean
    .withColumn("Sales", col("Sales").cast("double"))
    .withColumn("Quantity", col("Quantity").cast("int"))
)

df_clean.display()

```

```python
# Filtrar filas donde 'Sales' contenga solo números (enteros o decimales)
df_clean = df.filter(col("Sales").rlike(r"^[0-9]+(\.[0-9]+)?$"))

# Crear una nueva columna 'Ventas' como producto de 'Sales' * 'Quantity'
df_clean = df_clean.withColumn(
    "Ventas",
    col("Sales").cast("double") * col("Quantity").cast("double")
)
display(df_clean)
```

```python
ventas_por_subcategoria = df_clean.groupBy("Sub-Category").agg(sum(col("Quantity")).alias("Total_Ventas"))
display(ventas_por_subcategoria)
```
