```python
# Databricks Notebook: Python, Pandas & NumPy Basics for Data Analysis (2 Hours)
# Data source: workspace.trabajoenclase.superstore

# =====================================================
# 1. Introduction & Setup
# =====================================================
# Goals of this notebook:
# - Learn Databricks basics
# - Load Databricks table using Pandas
# - Understand Pandas & NumPy fundamentals
# - Perform basic data analysis and visualization

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# 2. Load Data using Pandas (from Spark Table)
# =====================================================
# Databricks tables are Spark tables, so we:
# 1. Read with Spark
# 2. Convert to Pandas for analysis

spark_df = spark.table("workspace.trabajoenclase.superstore")

# Convert to Pandas DataFrame
pdf = spark_df.toPandas()

pdf.head()
```

```python

# =====================================================
# 3. Pandas Basics
# =====================================================

# Shape of the dataset
pdf.shape

# Column names
pdf.columns

# Data types
pdf.dtypes

# Summary statistics
pdf.describe()

# Information about nulls and memory usage
pdf.info()

```

```python
# =====================================================
# 4. Basic Data Selection
# =====================================================

# Select one column
pdf["Sales"].head()
```

```python

# Select multiple columns
pdf[["Category", "Sales", "Profit"]].head()

```

```python

# Filter rows
pdf[pdf["Sales"] > 500].head()
```

```python

# =====================================================
# 5. Data Cleaning with Pandas
# =====================================================

# Check missing values
pdf.isna().sum()
```

```python

# Drop rows with missing Sales or Category
pdf_clean = pdf.dropna(subset=["Sales", "Category"])

# Ensure correct data types
pdf_clean["Sales"] = pdf_clean["Sales"].astype(float)
pdf_clean["Profit"] = pdf_clean["Profit"].astype(float)
pdf_clean["Quantity"] = pdf_clean["Quantity"].astype(int)

pdf_clean.dtypes
```

```python
x = "10"

x_int = int(x)
x_float = float(x)
x_str = str(x_int)

type(x_int)     # int
type(x_float)   # float
type(x_str)     # str
```

```python

# =====================================================
# 6. Pandas Aggregations & GroupBy
# =====================================================

# Total sales by category
sales_by_category = (
    pdf_clean
    .groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

sales_by_category

# Average profit by region
avg_profit_region = (
    pdf_clean
    .groupby("Region")["Profit"]
    .mean()
    .sort_values(ascending=False)
)

avg_profit_region
```

```python

# =====================================================
# 7. NumPy Basics
# =====================================================

# Create NumPy arrays
sales_array = np.array(pdf_clean["Sales"])
profit_array = np.array(pdf_clean["Profit"])

# Basic NumPy operations
sales_array.mean()
sales_array.max()
sales_array.min()

# Element-wise operations
profit_margin = profit_array / sales_array

profit_margin[:10]

# Boolean filtering with NumPy
high_margin = profit_margin > 0.3
high_margin.sum()

# =====================================================
# 8. Combining Pandas & NumPy
# =====================================================

# Add a NumPy-calculated column back to Pandas
pdf_clean["Profit_Margin"] = profit_margin

pdf_clean[["Sales", "Profit", "Profit_Margin"]].head()

# =====================================================
# 9. Visualization with Matplotlib
# =====================================================

sales_by_category.plot(kind="bar")
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python

# =====================================================
# 10. Mini Student Exercise
# =====================================================
# TASK:
# 1. Filter orders with Quantity > 5
# 2. Group by Sub_Category
# 3. Calculate average Sales
# 4. Sort descending

exercise = (
    pdf_clean[pdf_clean["Quantity"] > 5]
    .groupby("Sub-Category")["Sales"]
    .mean()
    .sort_values(ascending=False)
)

exercise.head()
```
